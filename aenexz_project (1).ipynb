**EDA**

How does attrition differ between employees who work overtime and those who don't?

```python
pd.crosstab(df["OverTime"], df["Attrition"])
```

**What does each number mean?**

Take this:

OverTime = No, Attrition = No → 944

That means:

944 employees did NOT work overtime AND stayed at the company.

And:

OverTime = Yes, Attrition = Yes → 127

means:

127 employees worked overtime AND left the company.


Overtime = Yes

        ↓

How many employees?

        ↓

How many left?

        ↓

What percentage left?

Then we'll compare it with:

Overtime = No

```python
attrition_rate_ot = pd.crosstab(
    df["OverTime"],
    df["Attrition"],
    normalize="index"
) * 100

print(attrition_rate_ot)
```

**MY OBSERVATION:**

I compared attrition rates rather than just raw employee counts. The attrition rate among overtime employees was 30.53%, compared with 10.44% among non-overtime employees. This indicates a strong association between overtime status and attrition in this dataset, although it doesn't establish causation.

**INTERPRETATION:**

Employees who work overtime have a 30.53% attrition rate, compared with 10.44% for employees who don't.

So the overtime group has an attrition rate about 2.9× higher.

```python
import matplotlib.pyplot as plt

# Select only employees who left
attrition_yes = attrition_rate_ot["Yes"]

# Create bar graph
plt.figure(figsize=(7, 5))
plt.bar(attrition_yes.index, attrition_yes.values)

plt.title("Attrition Rate by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Attrition Rate (%)")

# Show percentage on top of each bar
for i, value in enumerate(attrition_yes.values):
    plt.text(i, value + 1, f"{value:.2f}%", ha="center")

plt.ylim(0, 40)
plt.show()
```

**MONTHLY INCOME → ATTRITION**

```python
df["IncomeGroup"] = pd.cut(
    df["MonthlyIncome"],
    bins=4,
    labels=["Low", "Medium", "High", "Very High"]
)
```

MonthlyIncome

      ↓

   pd.cut()
   
      ↓
Low | Medium | High | Very High

```python
attrition_rate_ir = pd.crosstab(
    df["IncomeGroup"],
    df["Attrition"],
    normalize="index"
) * 100

print(attrition_rate_ir)
```

Low → 19.66%

    ↓

Medium → 14.20%

    ↓

High → 8.70%

    ↓

Very High → 3.79%


So in this dataset, employees in lower monthly-income groups have higher attrition rates than employees in higher-income groups.

The difference between the lowest and highest groups is:

19.66% − 3.79% = 15.87 percentage points

```python
# Get attrition rates
attrition_yes = attrition_rate_ir["Yes"]

# Create bar graph
plt.figure(figsize=(8, 5))

plt.bar(
    attrition_yes.index,
    attrition_yes.values
)

plt.title("Attrition Rate by Monthly Income Group")
plt.xlabel("Monthly Income Group")
plt.ylabel("Attrition Rate (%)")

# Add percentage labels
for i, value in enumerate(attrition_yes.values):
    plt.text(
        i,
        value + 0.5,
        f"{value:.2f}%",
        ha="center"
    )

plt.ylim(0, 25)
plt.show()
```

**MY OBSERVATION:**

I grouped employees into four monthly-income categories and compared their attrition rates. The low-income group had an attrition rate of 19.66%, while the very-high-income group had an attrition rate of only 3.79%. This shows a strong negative association between monthly income and attrition in the dataset, although it does not establish that income alone causes employees to leave.

**INTERPRETATION:**

Lower-income employees had considerably higher attrition. The rate decreased from 19.66% in the low-income group to 3.79% in the very-high-income group, showing a strong negative association between monthly income and attrition

**KPI 1 — Total Employees:**

```python
total_employees = df.shape[0]

print("Total Employees:", total_employees)
```

**KPI 2 — Employees Who Left:**

```python
employees_left = (df["Attrition"] == "Yes").sum()

print("Employees Who Left:", employees_left)
```

**KPI 3 — Employees Who Stayed**

```python
employees_stayed = (df["Attrition"] == "No").sum()

print("Employees Who Stayed:", employees_stayed)
```

**KPI 4 — Overall Attrition Rate**

```python
attrition_rate = (
    employees_left / total_employees
) * 100

print("Overall Attrition Rate:", round(attrition_rate, 2), "%")
```

**KPI 5 — Average Monthly Income**

```python
average_income = df["MonthlyIncome"].mean()

print("Average Monthly Income:", round(average_income, 2))
```

**KPI 6 — Average Age**

```python
average_age = df["Age"].mean()

print("Average Age:", round(average_age, 2))
```

**KPI 7 — Average Job Satisfaction**

```python
average_satisfaction = df["JobSatisfaction"].mean()

print(
    "Average Job Satisfaction:",
    round(average_satisfaction, 2)
)
```

**KPI 8 — Overtime Percentage**

```python
overtime_percentage = (
    (df["OverTime"] == "Yes").sum()
    / total_employees
) * 100

print(
    "Employees Working Overtime:",
    round(overtime_percentage, 2),
    "%"
)
```
